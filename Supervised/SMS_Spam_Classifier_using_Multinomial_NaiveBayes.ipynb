```python
import pandas as pd
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('/content/sms_spam_dataset_10k.csv')
df.head()
```

```python
df.info()
```

```python
X = df['message']
y = df['label']
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['message'], df['label'], test_size=0.25, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer = TfidfVectorizer()
```

```python
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

```python
print(X_train_tfidf.shape)
```

```python
print(vectorizer.get_feature_names_out())
```

```python
tfidf_df = pd.DataFrame(X_train_tfidf.toarray(), columns=vectorizer.get_feature_names_out())
tfidf_df.head()
```

```python
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)
```

```python
y_pred = model.predict(X_test_tfidf)
```

```python
from sklearn.metrics import accuracy_score, classification_report
print(accuracy_score(y_test, y_pred))
```

```python
print(classification_report(y_test, y_pred))
```

```python
new_message = ["Congratulations! You have won a free trip to Hawaii.",
               "Can we have a meeting.",
               "Claim your free prize now"
               ]
new_message_tfidf = vectorizer.transform(new_message)
prediction = model.predict(new_message_tfidf)
print(prediction)
```

```python
prob = model.predict_proba(new_message_tfidf)
print(prob)
```
